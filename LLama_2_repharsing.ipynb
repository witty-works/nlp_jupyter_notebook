```python
def json_extract(output):

     # Loop through the list to access each dictionary
     for record in output:
      generated_text = record.get('generated_text')
      # Find the index where the JSON object starts
      start_idx = generated_text.find('{')

      # Find the index where the JSON object ends
      end_idx = generated_text.rfind('}')

      # Extract the JSON object string
      json_str = generated_text[start_idx:end_idx + 1]

      # Parse the JSON string to a Python dictionary
      json_data = json.loads(json_str)

      # Extract the "sentence" field
      sentence = json_data.get('sentence', 'No sentence found')

      return sentence


```
